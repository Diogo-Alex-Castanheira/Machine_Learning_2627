#### 1.3.2 Identifier-like variables
The assignment asks us to treat identifier-like variables critically. We check whether `ID` and `customer_id` carry any signal about the target or merely encode ordering.

```python
# drops the 65 empty rows (they only have the IDs and the target, explained in Section 1.4)
train = train_raw[train_raw['Total Bruto'].notna()].copy()
for col in ['ID', 'customer_id']:
    # do purchases followed by a repurchase tend to have larger ID/customer_id numbers? (<0.5, tend to have smaller numbers | 0.5, it's a coin flip | >0.5, repeaters tend to have bigger numbers
    auc = roc_auc_score(train.repeat_purchase_90d, train[col])
    # do bigger ID/customer_id numbers go with later dates? (0 = no link with time | 1 = link with time)
    rho = train[col].corr(train.purchase_date.rank(), method='spearman') 
    print(f'{col:12s} AUC vs target = {auc:.3f} | Spearman with purchase date = {rho:.3f}')
```

**Insight:**
- **`ID`** tells us nothing about the target: its AUC is 0.51, which is a coin flip. It also does not follow time. The Spearman correlation with the purchase date is about 0 so `ID` is just a label.
- **`customer_id`** does not follow time either (Spearman ≈ −0.02). It has a weak lean toward the target (AUC 0.43, meaning purchases followed by a repurchase tend to have slightly smaller numbers). We cannot explain this from the data.
- Even a weak pattern in an identifier is a poor basis for prediction, because it does not describe customer behaviour. `customer_id` in particular would let flexible models memorise individual customers instead of learning patterns that apply to everyone.

**Decision:** `ID` is kept only as the submission key, and `customer_id` only as the key to join `clientes.csv`. Neither is used as a feature.

**C. History variables.**
- **First recorded purchases.** 603 train rows (9.3%) have no previous purchase recorded (`prior_purchase_count` = 0), versus 22 test rows (2.2%). We treat these as **first recorded purchases**: `customer_tenure_days` is 0 in all of them and in none of the other rows, each of the 603 customers has exactly one such row, and every one is dated on that customer's earliest purchase date. Test follows the same pattern (tenure 0 in all 22, and none of those customers appears in train). This does not exclude that a customer bought before the records start.
- **Empty recency.** `days_since_previous_purchase` is empty in 796 train rows (12.3%) and 52 test rows (5.3%). These are the 603 first recorded purchases plus **193 rows that are not first purchases**: they have previous purchases recorded (19 rows have 1, 20 have 2, 18 have 3, and so on, and the smallest previous purchase value is filled in for all 193) but no recency value.
- **Effect on the target.** First recorded purchases repurchase much less often (33.3%) than average. The 193 odd rows repurchase 57.0%, almost the same as rows with a recency value (58.3%).

**Consequence:**
- **First recorded purchases** (`prior_purchase_count` = 0) repurchase far less often (33%), so we create a flag for them. For these rows recency does not exist, so we fill it with the median learned from train and keep the flag.
- **The 193 rows with history but an empty recency** are not first purchases, and their repurchase rate is 57%, so the emptiness itself carries no signal. The value can be recomputed from the other previous purchase dates (at least for 192 of 193 of the rows). We recompute it instead of using the median.
- **Other empty history values** (for example `prior_mean_purchase_value` or `prior_mean_gap_days` with unexpected gaps) are filled with the median from the train set.

* **Logical inconsistencies:** 45 rows have a maximum single purchase above the total prior spend and 15 have `spend_180d` > `spend_365d`, which cannot both be true. They are rare (< 1%) but show that some historical aggregates are not fully reliable. Therefore, these variables should not be interpreted in isolation, and their usefulness will be assessed during feature selection and model validation.
* **Accounting identity:** `Net = Gross − discounts` holds exactly. `Total a Pagar` deviates from `Net + Tax + Shipping` However, in 1,053 rows, `Total a Pagar` does not equal Total Liquido + Tax + Shipping. All of these cases have a positive shipping charge, and the discrepancy is consistently negative (median −4.89). This suggests that the difference is related to the invoicing system's treatment of shipping rather than random data corruption. Since the exact business rule is unknown, we keep the original values unchanged rather than applying an arbitrary correction.
* **Selling below cost:** `Custo > Total Liquido` occurs in only 39 training rows (≈ 0.6%). In wholesale this can be returns, promotions or credit notes, so these rows are **kept**; log-transforming (Section 2) is a safer response than removal.

### 1.7. Numeric variables vs. target
Check which numerical variables are related to the target.

We use 3 different measures because they look at the relationship in different ways:

- **Spearman correlation:** checks whether the target tends to increase or decrease as the variable increases.
- **ANOVA F-test:** checks whether the variable has different average values for customers who repurchase and those who do not.
- **Mutual information:** can detect more complex relationships, including non-linear ones.

random_noise is included as a control variable. Since it was randomly generated, it should not contain useful information about the target. It therefore gives us a reference for what a variable with no real signal looks like.

```python
# Local copy for this analysis only: placeholders (1e9) and negative values are replaced by empty cells so that they do not
# distort the statistics (the real cleaning is done in Section 2). "money" is the list defined in 1.5.1
explore = train.copy()
for c in money:
    explore.loc[(explore[c] < 0) | (explore[c] >= 1e9), c] = np.nan
for c in ['days_since_previous_purchase', 'prior_mean_gap_days']:
    explore.loc[explore[c] < 0, c] = 0                                         # negative day gaps = same-day purchases

predictors = [c for c in num_cols if c not in ['ID', 'customer_id', 'repeat_purchase_90d']]    # numerical predictors only (no identifiers, no target)
y_train = explore.repeat_purchase_90d.values
X_filled = explore[predictors].fillna(explore[predictors].median())           # median fill ONLY here: the F-test and mutual information need complete data

F_stat, p_val = f_classif(X_filled, y_train)       # ANOVA F-test: do repeaters and non-repeaters have different average values?
univariate = pd.DataFrame({
    'spearman': explore[predictors].corrwith(explore.repeat_purchase_90d, method='spearman'),   # direction of the relationship
    'anova_F': F_stat, 'anova_p': p_val,
    'mutual_info': mutual_info_classif(X_filled, y_train, random_state=SEED)},                  # any kind of relationship
    index=predictors)
univariate['MI_rank'] = univariate.mutual_info.rank(ascending=False).astype(int)
pd.options.display.float_format = '{:,.4f}'.format
display(univariate.sort_values('mutual_info', ascending=False).head(15))
print('Control -> random_noise:', univariate.loc['random_noise'].round(4).to_dict(), '| AUC alone:', round(roc_auc_score(y_train, explore.random_noise), 3), '| out of', len(predictors), 'variables')

# Repurchase rate by bins of 4 variables (each bin holds about the same number of purchases)
def rate_by_bin(df, col, q=8):
    b = pd.qcut(df[col], q, duplicates='drop')
    return df.groupby(b, observed=True).repeat_purchase_90d.agg(['mean', 'size'])
explore['lateness_ratio'] = explore.days_since_previous_purchase / (explore.prior_mean_gap_days + 1)   # exploratory: is the customer late given their usual rhythm?
panels = ['days_since_previous_purchase', 'purchase_count_90d', 'prior_purchase_count', 'lateness_ratio']
fig, ax = plt.subplots(1, 4, figsize=(19, 3.8))
for a, c in zip(ax, panels):
    r = rate_by_bin(explore.assign(**{c: explore[c].clip(upper=explore[c].quantile(.99))}), c)   # the largest 1% of values are capped for readability
    a.plot(range(len(r)), r['mean'].values, marker='o'); a.set_xticks(range(len(r))); a.set_xticklabels([f'{i.left:.3g}–{i.right:.3g}' for i in r.index], rotation=55, fontsize=7)
    a.set_ylim(0.2, 1.0); a.set_title(c, fontsize=9); a.set_ylabel('repurchase rate')
plt.tight_layout(); plt.show()
```

**Insights**

* **The analysis uses a temporary cleaned copy of the data.** For this exploratory analysis, the artificial 1e9 sentinel values and negative monetary values are replaced with missing values so that they do not distort the statistics. Negative values in the time-gap variables are temporarily treated as 0. Missing values are then filled with the median only for the statistical tests that require complete data. The original dataset is not changed here; the final treatment of these values is done in Section 2.
* **Recency and frequency are the strongest signals.** Variables that describe how often and how recently a customer has purchased are the most clearly related to the target. In particular, `purchase_count_365d`, `purchase_count_180d`, `purchase_count_90d`, `prior_purchase_count`, and `prior_mean_gap_days` show relatively strong relationships with repurchasing. The plots support this: customers who purchase more frequently are more likely to repurchase, while customers who have gone longer without purchasing are less likely to do so.

* **The 90-day horizon is particularly interesting.** There is a noticeable drop in the repurchase rate around the 90-day mark (1st grafic). This makes sense because the target itself is whether the customer makes another purchase within 90 days. Therefore, variables describing recent purchases are expected to be especially informative.

* **Different statistical tests highlight different variables.** Spearman correlation and ANOVA mainly identify variables where the target changes in a relatively clear way between customers. Mutual information can also detect more complex relationships. For example, `prior_max_purchase_value` and `prior_min_purchase_value` rank highly according to mutual information even though their Spearman correlations are only moderate. This suggests that their relationship with the target may not be simply increasing or decreasing.

* **`random_noise` behaves as expected.** It has an AUC close to 0.5, a correlation close to 0, a non-significant ANOVA result, and very low mutual information. This is what we would expect from a variable containing no useful information. It is therefore used as a control to help us identify variables whose apparent signal may simply be due to chance.

* **The `lateness_ratio` provides a weaker signal.** This variable compares the number of days since the customer's last purchase with their usual time between purchases. The relationship with repurchasing is not as clear as the relationship using `days_since_previous_purchase` alone. Therefore, it is kept as a feature to test later, but we cannot yet say that it improves the model.

* **These tests are only a first screening.** A variable that looks weak on its own may still be useful when combined with other variables. Therefore, we do not remove variables based only on these results. Feature selection and model validation will be used later to decide which variables are actually useful.

### 1.9. Multivariate structure
Strongly correlated inputs are redundant (and unstable for linear models). Knowing which groups are collinear tells the feature-selection part where to look.

```python
# Data for the correlations: numerical predictors only (no identifiers, no target).
# Placeholders (1e9) and negative values are hidden so that they do not distort the correlations
# (`money` is the list defined in 1.5.1; the real cleaning is done in section 2)
corr_data = train[[c for c in num_cols if c not in ['ID', 'customer_id', 'repeat_purchase_90d']]].copy()
for c in money:
    corr_data.loc[(corr_data[c] < 0) | (corr_data[c] >= 1e9), c] = np.nan
for c in ['days_since_previous_purchase', 'prior_mean_gap_days']:
    corr_data.loc[corr_data[c] < 0, c] = 0                      # negative day gaps = same-day purchases

corr_matrix = corr_data.corr(method='spearman')                 # empty cells are ignored pair by pair

# Heatmap: only the lower triangle is shown, so each pair appears once; colours go from -1 to +1
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
fig, ax = plt.subplots(figsize=(11, 8.5))
sns.heatmap(corr_matrix, mask=mask, cmap='coolwarm', center=0, vmin=-1, vmax=1, ax=ax, cbar_kws={'shrink': .7})
ax.set_title('Spearman correlation between numerical predictors'); plt.tight_layout(); plt.show()

# List every pair of variables once (upper triangle of the matrix) with its correlation
upper = np.triu_indices_from(corr_matrix, k=1)
pairs = pd.DataFrame({'variable_a': corr_matrix.index[upper[0]], 'variable_b': corr_matrix.columns[upper[1]], 'rho': corr_matrix.values[upper]})
pairs['abs_rho'] = pairs.rho.abs()                              # strength of the relationship, ignoring the sign

# How many pairs are strongly correlated?
print({f'|rho|>{t}': int((pairs.abs_rho > t).sum()) for t in (0.95, 0.9, 0.8)}, 'out of', len(pairs), 'pairs')
top_pairs = pairs[pairs.abs_rho > 0.80].sort_values('abs_rho', ascending=False)
print('Pairs with |rho| > 0.80 (top 20):')
display(top_pairs.head(20).round(3))
```

**Insights**

* **Most numerical variables are not strongly correlated.** Only 4 pairs have \(|ρ| > 0.95\) and 7 have \(|ρ| > 0.90\), so redundancy is concentrated in a few groups.
* **The strongest redundancy is among the transaction-value variables:** Total a Pagar, Total Liquido and Total Bruto have correlations above 0.99. These variables contain very similar information.
* **There is also strong correlation between purchase counts and spending** over the same period, for example purchase_count_30d and spend_30d (ρ = 0.97), and between historical purchase measures such as prior_purchase_count and prior_total_spend (ρ = 0.90).
* **The recent purchase behavoiur variables are also correlated with each other.** This is expected because, for example, purchases in the last 90 days are also part of the last 180 and 365 days.
* **For feature selection:** we should remove or select between highly redundant variables rather than keeping all of them. A threshold around \(|ρ| > 0.90\) is a reasonable starting point, although the final choice should be checked using model performance.

We use **Spearman** correlation because the numerical variables are often skewed and contain extreme values. Spearman measures whether two variables tend to increase/decrease together without assuming a linear relationship.

### 1.10. Train vs. test: does the test set look like the training set?
A model can only transfer to test if the inputs have similar distributions. We compare each numeric variable with the two-sample Kolmogorov–Smirnov statistic (0 = identical distributions, 1 = completely different). Only inputs are compared; the target is not available in test.

```python
# Numerical predictors = numeric columns that exist in BOTH files (test has no target and no helper columns like `year`), without identifiers
predictor_cols = [c for c in num_cols if c in test_raw.columns and c not in ['ID', 'customer_id']]

# Placeholders (1e9) and negative values are hidden in BOTH files, so that they do not distort the comparison
# (`money` is the list defined in 1.5.1; the real cleaning is done in section 2)
def hide_implausible(df):
    d = df.copy()
    for c in money:
        d.loc[(d[c] < 0) | (d[c] >= 1e9), c] = np.nan
    for c in ['days_since_previous_purchase', 'prior_mean_gap_days']:
        d.loc[d[c] < 0, c] = 0                                   # negative day gaps = same-day purchases
    return d
train_view, test_view = hide_implausible(train), hide_implausible(test_raw)

# For each variable: KS statistic between train and test, plus the median in each file
rows = []
for c in predictor_cols:
    a, b = train_view[c].dropna(), test_view[c].dropna()
    rows.append((c, ks_2samp(a, b).statistic, a.median(), b.median()))
shift = pd.DataFrame(rows, columns=['variable', 'KS', 'median_train', 'median_test']).set_index('variable').sort_values('KS', ascending=False)
display(shift.head(10).round(3))
print('Median KS over all variables:', round(shift.KS.median(), 3))
print('random_noise (reference for "no drift"):', shift.loc['random_noise'].round(3).to_dict())
```

### Insights

* **Some variables are noticeably different between train and test.** The largest differences are found in `customer_tenure_days`, `prior_total_spend`, `prior_purchase_count`, and other variables that describe a customer's accumulated history.

* **These differences are expected because the test period is later in time.** As time passes, customers can accumulate more purchases, spend more money, and have longer relationships with the company. For example, the median `customer_tenure_days` increases from 918 days in train to 2,662 days in test.

* **The test set therefore represents a somewhat different population from the training set.** This is an important limitation because the model needs to make predictions for customers whose historical values can be larger than those commonly seen during training.

* **This reinforces the need for temporal validation.** A random train/validation split could hide these time-related differences. A temporal split is more realistic because it allows us to test whether a model trained on earlier data can predict observations from a later period.

* **The median KS statistic is 0.127.** This suggests that, although some variables show noticeable distribution shifts, the difference is not equally strong across all numerical variables.

* **We therefore keep these variables for now rather than removing them automatically.** Their usefulness will be evaluated during feature selection and model validation, using a validation setup that respects the temporal structure of the data.

## 2. Data Cleaning and Preprocessing

In Section 1, we identified missing purchase information, anomalies in historical variables, rare categorical levels, and differences between the training and test periods. We now use these findings to prepare the data for classification. We keep the cleaning rules separate from transformations that learn parameters, so the latter can be fitted only on the training observations in each validation split.

### 2.1. Missing values and observations to retain

We start again from the original data, rather than from the temporary copies used for EDA. As discussed in Section 1.4, **65 training observations** contain only identifiers and the target, so we remove them. We retain purchases with partial missing values and all test observations. We also keep repeated same-day purchases: these may represent separate purchase occasions rather than erroneous duplicates.

```python
from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

target = 'repeat_purchase_90d'
# Section 1 already loaded train_raw, test_raw and clientes.
transaction_fields = [c for c in train_raw if c not in ['ID', 'customer_id', target]]
empty_purchase = train_raw[transaction_fields].isna().all(axis=1)
train_clean = train_raw.loc[~empty_purchase].copy()
print('Rows removed:', empty_purchase.sum())
print('Training rows retained:', len(train_clean), '| test rows retained:', len(test_raw))
assert empty_purchase.sum() == 65
assert train_clean['ID'].is_unique and test_raw['ID'].is_unique
```

**Treatment of missing information.** We distinguish the meaning of missing values before deciding how to handle them. The original null counts are discussed in Section 1; here we focus on the action that follows from them.

| Situation identified in Section 1 | Treatment in our pipeline | Reason |
|---|---|---|
| Almost entirely empty transaction (65 training rows) | Remove from labelled training only | No usable transaction information remains. |
| No previous purchase (`prior_purchase_count = 0`) | Retain a first-purchase flag and leave undefined historical statistics missing | A first purchase does not have a previous purchase gap; filling with zero would suggest a zero-day gap. |
| Recency missing despite previous purchases | Attempt reconstruction from earlier observed purchase dates; impute unresolved cases | Historical gaps may be recoverable, but never from a future transaction. |
| Invalid numeric history (`1e9`, negative where impossible) | Replace with `NaN`, retain invalid-value indicator | The value cannot safely be treated as an actual amount or interval. |
| Remaining missing numeric values | Training-fold **median** with missingness indicators | Several amounts are skewed, making the median less sensitive to very large purchases than the mean. |
| Missing categorical values | Explicit `Missing` category | We do not assume missing categories equal the most common level. |

The median and other learned quantities are estimated **only from the training partition**. We use these choices as our first consistent baseline, not as evidence that alternative imputers could not perform better.

### 2.2. Integrating the customer table

Section 1.11 showed that `clientes.csv` contains one record per customer and covers the purchases in both files. We therefore add a small set of customer attributes through a **many-to-one left join**. We keep broad geographical information and the year the customer relationship began; we do not include addresses or near-identifiers. The derived customer age is calculated relative to each purchase date, not to the present day. We will leave the question of whether these attributes improve prediction to the modelling stage.

```python
customer_fields = ['customer_id', 'country_or_region', 'commercial_region',
                   'island', 'customer_since_year']
customer_lookup = clientes[customer_fields].copy()
assert customer_lookup['customer_id'].is_unique

def join_customer_data(df):
    result = df.merge(customer_lookup, on='customer_id', how='left',
                      validate='many_to_one', indicator=True, sort=False)
    assert (result['_merge'] == 'both').all()
    assert len(result) == len(df)
    return result.drop(columns='_merge')

train_joined = join_customer_data(train_clean)
test_joined = join_customer_data(test_raw)
print('Customer match rate: 100%; rows unchanged after join.')
```

### 2.3. Treating anomalies and creating candidate features

We use fixed rules for clear data-quality problems: artificial `1e9` historical values and negative quantities where they are not meaningful become missing. We **do not** automatically delete extreme purchase amounts, correct accounting totals, or remove rare historical inconsistencies, because Section 1 could not establish that those observations were erroneous. Instead, we add indicators for selected inconsistencies.

For missing recency in non-first purchases, we first look for an earlier dated purchase of the same customer **in the available training history**. This reconstructs recency only where an earlier purchase is actually observed; otherwise it remains missing for later train-fitted imputation. We preserve the first-purchase flag and also identify recovered recency. For validation and test, the lookup is fitted on the earlier training period only, rather than learning anything from `test.csv`.

We also examine a few simple derived features related to the 90-day repeat-purchase question. These are **candidate features**, not a claim that all improve the final model.

To avoid discarding potentially valid large orders, we do **not** remove observations using an arbitrary IQR or z-score threshold. For ratio and log features, we handle undefined denominators as missing and use `log1p` only for nonnegative inputs. Feature usefulness, including correlated rolling totals, remains a question for Section 3.

```python
bad_history = ['prior_total_spend', 'prior_mean_purchase_value',
               'prior_std_purchase_value', 'prior_min_purchase_value',
               'prior_max_purchase_value', 'spend_30d', 'spend_90d',
               'spend_180d', 'spend_365d']
nonneg_timing = ['days_since_previous_purchase', 'customer_tenure_days',
                 'prior_mean_gap_days', 'prior_std_gap_days']
count_columns = ['N_ Detalhes', 'invoice_count_current_purchase',
                 'prior_purchase_count', 'purchase_count_30d',
                 'purchase_count_90d', 'purchase_count_180d',
                 'purchase_count_365d']
customer_categoricals = ['country_or_region','commercial_region','island']

class PurchaseFeatureBuilder(BaseEstimator, TransformerMixin):
    """Fixed row-wise rules; previous dates come only from the data passed to fit."""
    def __init__(self, include_customers=True, include_noise=True):
        self.include_customers = include_customers
        self.include_noise = include_noise

    def fit(self, X, y=None):
        dates = X[['customer_id', 'purchase_date']].copy()
        dates['purchase_date'] = pd.to_datetime(dates['purchase_date'], errors='coerce')
        self.observed_dates_ = {
            key: np.sort(group['purchase_date'].dropna().unique().astype('datetime64[ns]'))
            for key, group in dates.groupby('customer_id')
        }
        return self

    def transform(self, X):
        d = X.copy()
        date = pd.to_datetime(d['purchase_date'], errors='coerce')
        d['is_first_purchase'] = d['prior_purchase_count'].eq(0).astype(int)
        d['purchase_month_sin'] = np.sin(2*np.pi*(date.dt.month-1)/12)
        d['purchase_month_cos'] = np.cos(2*np.pi*(date.dt.month-1)/12)
        d['purchase_day_of_week'] = date.dt.dayofweek.astype(float)

        for c in bad_history:
            invalid = d[c].lt(0) | d[c].ge(1e9)
            d[c+'_invalid_flag'] = invalid.astype(int)
            d[c] = d[c].mask(invalid)
        for c in nonneg_timing + count_columns:
            invalid = d[c].lt(0)
            d[c+'_invalid_flag'] = invalid.astype(int)
            d[c] = d[c].mask(invalid)

        d['recency_reconstructed_flag'] = 0
        needs_recency = d['days_since_previous_purchase'].isna() & d['prior_purchase_count'].gt(0)
        # Never borrow a date on the same day or after the current purchase.
        for idx in d.index[needs_recency]:
            customer_dates = self.observed_dates_.get(d.at[idx, 'customer_id'])
            current_date = np.datetime64(date.loc[idx], 'ns') if pd.notna(date.loc[idx]) else np.datetime64('NaT')
            if customer_dates is None or np.isnat(current_date):
                continue
            pos = np.searchsorted(customer_dates, current_date, side='left')
            if pos > 0:
                recency = (current_date - customer_dates[pos-1]) / np.timedelta64(1, 'D')
                d.at[idx, 'days_since_previous_purchase'] = float(recency)
                d.at[idx, 'recency_reconstructed_flag'] = 1

        d['prior_max_exceeds_total_flag'] = (
            d['prior_max_purchase_value'] > d['prior_total_spend'] + 1).astype(int)
        d['rolling_spend_inconsistent_flag'] = (
            (d['spend_90d'] > d['spend_180d'] + 1) |
            (d['spend_180d'] > d['spend_365d'] + 1)).astype(int)
        d['rolling_count_inconsistent_flag'] = (
            d['purchase_count_90d'] > d['purchase_count_365d']).astype(int)
        d['current_profit_margin'] = (d['Total Liquido'] - d['Custo']) / d['Total Liquido'].replace(0, np.nan)
        d['current_discount_rate'] = (
            d['Total Desc_ Global'] + d['Total Desc_ Linha']) / d['Total Bruto'].replace(0, np.nan)
        d['avg_current_invoice_value'] = (
            d['Total Liquido'] / d['invoice_count_current_purchase'].replace(0, np.nan))
        d['past_90d_purchase_rate'] = d['purchase_count_90d'] / 90.0
        d['past_90d_avg_spend'] = d['spend_90d'] / d['purchase_count_90d'].replace(0, np.nan)
        d['lateness_ratio'] = (d['days_since_previous_purchase'] /
                               d['prior_mean_gap_days'].replace(0, np.nan))
        for c in ['prior_total_spend', 'spend_30d', 'spend_90d', 'spend_180d',
                  'spend_365d', 'prior_mean_purchase_value']:
            d['log_'+c] = np.log1p(d[c].clip(lower=0))
        if self.include_customers:
            d['customer_age_at_purchase'] = (date.dt.year - pd.to_numeric(
                d['customer_since_year'], errors='coerce'))
            d['customer_age_at_purchase'] = d['customer_age_at_purchase'].mask(
                (d['customer_age_at_purchase'] < 0) | (d['customer_age_at_purchase'] > 150))
        else:
            d = d.drop(columns=customer_categoricals, errors='ignore')
        drop = ['ID', 'customer_id', 'purchase_date', 'customer_since_year', target]
        if not self.include_noise:
            drop.append('random_noise')
        result = d.drop(columns=drop, errors='ignore')
        return result.replace([np.inf, -np.inf], np.nan)

# Audit counts on training observations before any imputation.
for col in ['prior_mean_purchase_value', 'spend_365d', 'days_since_previous_purchase']:
    print(col, '| negative:', int(train_joined[col].lt(0).sum()),
          '| >= 1e9:', int(train_joined[col].ge(1e9).sum()),
          '| missing:', int(train_joined[col].isna().sum()))
```

### 2.4. Encoding, rare categories and scaling

Section 1.8 showed that `transport_method` has many rare levels. We combine levels appearing fewer than **10 times in the training fold** into `"Other"`, then one-hot encode the categorical features. Unseen categories are handled by the encoder rather than causing an error. Missing categories are represented as `"Missing"`.

For numerical features, we use a **training-fitted median** for remaining missing values and standardise using the training-fold mean and standard deviation. This is appropriate for the logistic-regression baseline and other scale-sensitive models; tree models would not normally require scaling. The imputer, rare-level grouping and scaler are all fitted inside the pipeline, so validation and test observations cannot determine their parameters.

**Encoding choice.** We start with one-hot encoding because payment methods, warehouses and transport methods represent labels, not ranked categories. Ordinal encoding would impose an artificial ordering. Count or target encoding could be investigated later, but target-based encoders need particular care to avoid fitting on the validation labels.

**Scaling choice.** StandardScaler is our starting choice for the logistic-regression baseline; a tree-based model typically does not need it. In Section 4 we can compare the baseline with no scaling or RobustScaler where appropriate, using the same validation design rather than asserting that one scaler is universally best.

```python
class RareCategoryGrouper(BaseEstimator, TransformerMixin):
    """Group rare transport categories using frequencies from fit data only."""
    def __init__(self, min_count=10):
        self.min_count = min_count

    def fit(self, X, y=None):
        x = pd.DataFrame(X).copy().fillna('Missing').astype(str)
        counts = x.iloc[:, 0].value_counts()
        self.frequent_ = set(counts[counts >= self.min_count].index)
        self.frequent_.add('Missing')
        return self

    def transform(self, X):
        x = pd.DataFrame(X).copy().fillna('Missing').astype(str)
        return np.where(x.iloc[:, 0].isin(self.frequent_),
                        x.iloc[:, 0], 'Other').reshape(-1, 1)

# Column types are taken from the labelled training schema, not test.csv.
feature_builder = PurchaseFeatureBuilder(include_customers=True, include_noise=True)
feature_builder.fit(train_joined.drop(columns=target))
sample_features = feature_builder.transform(train_joined.drop(columns=target))
cat_cols = sample_features.select_dtypes(include=['object', 'category', 'string']).columns.tolist()
num_cols = sample_features.select_dtypes(include=np.number).columns.tolist()
regular_cats = [c for c in cat_cols if c != 'transport_method']

try:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

numeric_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='median', add_indicator=True)),
    ('scale', StandardScaler())
])
regular_cat_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='constant', fill_value='Missing')),
    ('encode', clone(encoder))
])
transport_pipeline = Pipeline([
    ('group_rare', RareCategoryGrouper(min_count=10)),
    ('encode', clone(encoder))
])
column_steps = [('numeric', numeric_pipeline, num_cols)]
if regular_cats:
    column_steps.append(('categorical', regular_cat_pipeline, regular_cats))
if 'transport_method' in cat_cols:
    column_steps.append(('transport', transport_pipeline, ['transport_method']))

preprocessing = Pipeline([
    ('features', PurchaseFeatureBuilder(include_customers=True, include_noise=True)),
    ('columns', ColumnTransformer(column_steps, remainder='drop'))
])
print('Candidate numeric columns:', len(num_cols), '| categorical columns:', len(cat_cols))
print('We retain random_noise as a negative control for feature selection.')
```

### 2.5. Checking the preprocessing on a chronological split

As we noted in Section 1.3, purchases from the same customer appear more than once, and the test period comes later. We therefore split the retained training data by purchase date, using earlier observations for training and later observations for validation. Here we demonstrate that preprocessing fitted on the earlier period can transform the later period without learning its imputation values, category frequencies or scaling parameters. The modelling section will use the same principle when comparing models and selecting features.

**Recency reconstruction:** We reconstruct missing recency values using earlier purchase dates from the available training history. Same-day and future purchases are excluded to prevent data leakage. When there is insufficient historical information, the value remains missing and is handled by the imputation pipeline.

```python
ordered = train_joined.sort_values(['purchase_date', 'ID']).reset_index(drop=True)
cut = int(len(ordered) * 0.8)
boundary = ordered.loc[cut, 'purchase_date']
earlier = ordered.loc[ordered['purchase_date'] < boundary].copy()
later = ordered.loc[ordered['purchase_date'] >= boundary].copy()
assert earlier['purchase_date'].max() < later['purchase_date'].min()
X_train, y_train = earlier.drop(columns=target), earlier[target]
X_valid, y_valid = later.drop(columns=target), later[target]

fitted_preprocessing = clone(preprocessing).fit(X_train, y_train)
prepared_train = fitted_preprocessing.transform(X_train)
prepared_valid = fitted_preprocessing.transform(X_valid)
assert prepared_train.shape[1] == prepared_valid.shape[1]
assert np.isfinite(prepared_train).all() and np.isfinite(prepared_valid).all()
print('Train / validation rows:', len(X_train), len(X_valid))
print('Transformed train / validation shapes:', prepared_train.shape, prepared_valid.shape)

# Show how many missing recency values can be recovered from the earlier training history.
builder_fitted = fitted_preprocessing.named_steps['features']
for label, frame in [('Train', X_train), ('Validation', X_valid)]:
    output = builder_fitted.transform(frame)
    print(label, 'recency reconstructed:', int(output['recency_reconstructed_flag'].sum()))

# A diagnostic baseline only; model selection is addressed in Section 4.
baseline = Pipeline([
    ('preprocessing', clone(preprocessing)),
    ('model', LogisticRegression(max_iter=2000, random_state=SEED))
])
baseline.fit(X_train, y_train)
valid_scores = baseline.predict_proba(X_valid)[:, 1]
print('Illustrative validation ROC-AUC:', round(roc_auc_score(y_valid, valid_scores), 4))
```

We use median imputation, one-hot encoding, and StandardScaler as our initial preprocessing approach. These choices provide a consistent baseline for modelling, while alternative methods can be evaluated during model optimization.

### 2.6. Summary and hand-off to feature selection

Our preprocessing reflects the EDA decisions: we remove the 65 empty training observations, keep valid transactions and same-day purchases, replace clear historical errors with missing values, and retain accounting amounts that we cannot justify correcting. We distinguish first recorded purchases from other missing recency observations, try to recover recency using earlier known purchases, and use train-fitted imputation for the remaining gaps. We join selected customer attributes, combine rare transport categories, encode categorical fields, and scale the numerical features for the initial linear model.

We keep `random_noise` in the candidate feature set **as a negative control** and leave the decision to remove it, along with `document_series`, correlated monetary variables and any unhelpful engineered features, to Section 3. We also leave the customer-data ablation and the final choice of classification metrics/models to the next stage. All learned transformations must be refitted within each training split and, after model selection, fitted on the entire retained `train.csv` only. `test.csv` is transformed with that fitted pipeline; `ID` is kept outside the predictor matrix for the eventual Kaggle submission.

We therefore regard the proposed imputer, encoder and scaler as **initial choices for evaluation**, not as conclusively optimal settings. In particular, Section 3 can test whether the engineered variables are informative, and Section 4 can compare appropriate alternative preprocessing configurations under the same chronological assessment.

**Preprocessing decision log**

| EDA finding | Chosen action | Why we chose it |
|---|---|---|
| 65 rows without usable transaction attributes | Remove only these training rows | Very limited information for modelling |
| Some customer histories are missing or incomplete | First-purchase flag, recover past-only recency where possible, median imputation and indicators | Missing can mean genuinely absent history or unavailable measurements |
| Sentinel and implausible historical values | Turn invalid numbers into missing, preserve anomaly flags | Avoid teaching the model false numerical magnitudes |
| Unusual but potentially legitimate transactions | Keep transaction values | Avoid unsupported outlier deletion |
| Multiple customers' categories and rare transport levels | Explicit missing category, training-fitted rare-grouping, one-hot encoding | Represent categories without false rank or exploding rare levels |
| `clientes.csv` has unique `customer_id` | Left join selected customer attributes | Adds contextual features without duplicating transactions |
| Heavy right skew in historical spend | Add nonnegative `log1p` candidates | Compress large values without deleting transactions |
| Numeric features have different scales | Training-fitted StandardScaler for linear baseline | Makes magnitude more comparable for scale-sensitive methods |
| `random_noise` is a negative control | Keep as a candidate for feature selection | Test whether feature selection rejects uninformative signals |
| Later purchase occasions form the held-out period | Chronological validation, train-only fitting | Reduce temporal leakage and reflect prediction of later purchases |

## 3. Feature Selection

Section 2 produced **79 candidate features** (70 numeric, 9 categorical): the original variables, the selected customer attributes, the engineered features, the invalid-value flags and `random_noise`. Many of them carry the same information (e.g. `spend_365d` and `log_spend_365d`) or almost none. Here we reduce them to a smaller set that keeps the predictive signal, is easier to explain and is less likely to overfit (finding #22: small dataset, many rows per customer).

**Strategy.** We combine the three families of methods seen in class, each answering a different question:

| Step | Method | Type | Question it answers |
|---|---|---|---|
| 3.2 | Critical review of identifiers | Domain knowledge | Should this variable be a predictor at all? |
| 3.3 | Near-constant check | Filter (no target) | Does the feature vary enough to be learned? |
| 3.4 | Spearman / Cramér's V redundancy | Filter (no target) | Does another feature already carry the same information? |
| 3.5 | L1-regularised logistic regression | Embedded (linear) | Which features does a sparse linear model keep? |
| 3.6 | Permutation importance vs. `random_noise` | Model-agnostic (non-linear) | How much does each feature help predict **later** purchases, compared with pure noise? |
| 3.7 | Backward elimination guided by the importance ranking | Wrapper | What is the smallest set that predicts as well as the larger sets? |
| 3.8 | Ablations | Wrapper | Do the customer block, calendar features and other open questions really help? |

**No leakage, same validation logic as Section 2.** All selection work uses only the **earlier 80% of the training period** (`X_train` from 2.5). Inside it we use **4 expanding time folds**: train on everything before a date, validate on the next block of dates. The **later 20%** (`X_valid`) is not touched here, so Section 4 can still give an honest final estimate after the features have been chosen. The feature builder, imputer, encoder and scaler are refitted inside every fold.

**Two model families.** Linear and tree-based models "see" features differently (e.g. a tree can use a non-linear effect that a linear model misses). To avoid a feature set that only suits one model, we check every decision with a logistic regression (LR) and a shallow gradient-boosting model (HGB). We use **ROC-AUC** as the main metric and Average Precision as a second one (finding #12).

### 3.1. Setup: data, time folds and helper functions

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import average_precision_score
from scipy.stats import chi2_contingency

# we only use the earlier 80% here, X_valid stays untouched for section 4
X_fs = X_train.sort_values(['purchase_date', 'ID']).reset_index(drop=True)
y_fs = y_train.loc[X_train.sort_values(['purchase_date', 'ID']).index].reset_index(drop=True)

fs_num = list(num_cols)
fs_cat = list(cat_cols)
all_candidates = fs_num + fs_cat
print('Rows:', len(X_fs), '| candidate features:', len(all_candidates))

# 4 time folds: train on all dates before a cut-off, validate on the next block of dates
dates = X_fs['purchase_date']
cuts = dates.quantile([0.4, 0.55, 0.7, 0.85, 1.0]).values
folds = []
for i in range(4):
    tr = np.where(dates < cuts[i])[0]
    if i < 3:
        va = np.where((dates >= cuts[i]) & (dates < cuts[i + 1]))[0]
    else:
        va = np.where(dates >= cuts[i])[0]
    folds.append((tr, va))
    print(f'Fold {i+1}: train {len(tr)} rows (until {dates[tr].max().date()}), '
          f'valid {len(va)} rows ({dates[va].min().date()} to {dates[va].max().date()}), '
          f'repeat rate {y_fs[va].mean():.3f}')

# same preprocessing as section 2, but only on the features we give it
def make_pipe(features, model):
    num = [f for f in features if f in fs_num]
    cat = [f for f in features if f in fs_cat and f != 'transport_method']
    steps = []
    if num:
        steps.append(('numeric', clone(numeric_pipeline), num))
    if cat:
        steps.append(('categorical', clone(regular_cat_pipeline), cat))
    if 'transport_method' in features:
        steps.append(('transport', clone(transport_pipeline), ['transport_method']))
    return Pipeline([('columns', ColumnTransformer(steps)), ('model', model)])

# build the features once per fold (the builder is fitted on the training part only)
fold_data = []
for tr, va in folds:
    fb = PurchaseFeatureBuilder().fit(X_fs.iloc[tr])
    fold_data.append((fb.transform(X_fs.iloc[tr]), y_fs.iloc[tr], fb.transform(X_fs.iloc[va]), y_fs.iloc[va]))

def cv_folds(features, model):
    aucs, aps = [], []
    for F_tr, y_tr, F_va, y_va in fold_data:
        pipe = make_pipe(features, clone(model)).fit(F_tr, y_tr)
        prob = pipe.predict_proba(F_va)[:, 1]
        aucs.append(roc_auc_score(y_va, prob))
        aps.append(average_precision_score(y_va, prob))
    return np.array(aucs), np.array(aps)

lr_model = LogisticRegression(C=0.1, max_iter=5000, random_state=SEED)
hgb_model = HistGradientBoostingClassifier(max_depth=3, learning_rate=0.05, max_iter=300, random_state=SEED)

# features on the whole period, only used by the filters (they don't look at the target)
F_fs = PurchaseFeatureBuilder().fit(X_fs).transform(X_fs)
```

**Why this setup:**
- **Folds split on dates, not rows.** A date never appears in both the training and validation part of a fold, so same-day purchases of one customer cannot leak across the split.
- **Expanding window.** Each fold trains on all earlier purchases (2,068 → 4,392 rows) and validates on the next ~775 purchases, which mimics how the model will be used on `test.csv` (later purchases).
- **The repeat rate varies by fold (0.51–0.58).** Results also vary between folds, so we always compare feature sets **on the same folds** (paired comparison) rather than looking only at averages.
- **Selection is done on original features, not on one-hot columns.** `make_pipe` keeps the Section 2 preprocessing and only changes which features go in. A categorical variable is kept or dropped as a whole.
- **Library versions.** The results below were produced with scikit-learn 1.6.1, pandas 2.2.3 and numpy 2.1.3 (see `requirements.txt`). Other versions can slightly change the small importance values in 3.6, so the whole group should use the same versions.

### 3.2. Identifier-like variables

Section 1.3.2 showed that `ID` has no relation with the target (AUC 0.51) and that `customer_id` has a weak pattern we cannot explain (AUC 0.43). An identifier does not describe customer behaviour: `customer_id` would let a flexible model memorise individual customers, and the weak pattern could disappear or flip for new customers. We therefore **exclude both** (finding #3). `ID` is only kept as the submission key and `customer_id` only as the join key for `clientes.csv`.

The same logic applies to `purchase_date` (raw date = position in time, not behaviour; its useful parts are already in the recency and calendar features) and to `customer_since_year` (replaced by `customer_age_at_purchase`). Section 2's feature builder already removes them; we check this below.

We **keep `random_noise` for now** as a negative control: it is random by construction, so any feature that does not beat it has no reliable signal (finding #14).

```python
for col in ['ID', 'customer_id', 'purchase_date', 'customer_since_year', target]:
    assert col not in all_candidates
print('IDs, date and target are not in the candidates')
print('random_noise still in the candidates:', 'random_noise' in all_candidates)
```

### 3.3. Filter 1 – near-constant features

A feature that has the same value in (almost) every row cannot help the model separate the two classes, and the few different rows are too rare to learn from reliably. We drop features whose most frequent value covers **99% or more** of the rows. This filter does not use the target.

```python
# share of the most common value in each feature
top_share = pd.Series({c: F_fs[c].fillna('Missing').value_counts(normalize=True).iloc[0] for c in all_candidates})
near_constant = top_share[top_share >= 0.99].sort_values(ascending=False)

display(pd.DataFrame({'top_value_share': near_constant.round(4),
                      'rows_different': ((1 - near_constant) * len(F_fs)).round().astype(int)}))

after_nc = [c for c in all_candidates if c not in near_constant.index]
print('Removed:', len(near_constant), '| left:', len(after_nc))
```

**Insights**
- **23 features are removed, all of them flags created in Section 2.3.**
- **18 flags are always 0** in the feature-selection period. The anomaly they check for (e.g. a negative `purchase_count_90d`) never happens there, so they carry no information.
- **5 flags fire in only 12–39 rows** (0.2%–0.8%). Too few to learn a reliable effect. The underlying problem is still handled: the invalid values were set to missing in Section 2, and the imputer's missing-value indicators still mark those rows.
- `is_first_purchase` (9% of rows) and the categorical variables with one dominant level (`commercial_zone` 97%, `warehouse` 97%, `country_or_region` 97%) stay below the threshold. They are judged by the next steps.

### 3.4. Filter 2 – redundant features

Section 1.9 found that the redundancy sits in a few groups (finding #18). Keeping both members of a highly correlated pair adds no information, makes linear coefficients unstable and splits importance between the two. Following the EDA, we use **Spearman |ρ| > 0.90** as the threshold (Spearman because the variables are skewed).

**Rule:** sort the numeric features from strongest to weakest univariate ROC-AUC. Go down the list and keep a feature only if it is not > 0.90 correlated with one already kept. When two features are tied (a variable and its `log1p` version have identical rank information), we keep the log version because it suits scale-sensitive models better. For categorical variables we check pairs with Cramér's V.

```python
num_left = [c for c in after_nc if c in fs_num]
rho = F_fs[num_left].corr(method='spearman').abs()

# univariate AUC, only used to choose which feature of a pair we keep
filled = F_fs[num_left].fillna(F_fs[num_left].median())
uni_auc = {}
for c in num_left:
    auc = roc_auc_score(y_fs, filled[c])
    uni_auc[c] = max(auc, 1 - auc)

# strongest first; if two are tied we prefer the log version
order = sorted(num_left, key=lambda c: (-round(uni_auc[c], 4), not c.startswith('log_')))
kept_num = []
dropped = []
for c in order:
    too_close = [k for k in kept_num if rho.loc[c, k] > 0.90]
    if too_close:
        k = too_close[0]
        dropped.append([c, k, round(rho.loc[c, k], 3), round(uni_auc[c], 3), round(uni_auc[k], 3)])
    else:
        kept_num.append(c)

redundant = pd.DataFrame(dropped, columns=['dropped', 'kept_instead', 'spearman_rho', 'auc_dropped', 'auc_kept'])
display(redundant)

# same idea for categorical variables, with Cramer's V
def cramers_v(a, b):
    table = pd.crosstab(a, b)
    chi2 = chi2_contingency(table)[0]
    return np.sqrt(chi2 / (table.values.sum() * (min(table.shape) - 1)))

cat_left = [c for c in after_nc if c in fs_cat]
for i, a in enumerate(cat_left):
    for b in cat_left[i + 1:]:
        v = cramers_v(F_fs[a].fillna('Missing'), F_fs[b].fillna('Missing'))
        if v > 0.8:
            print(f"Cramer's V {a} / {b}: {v:.3f}")

after_filters = kept_num + cat_left
print('Numeric kept:', len(kept_num), '| categorical kept:', len(cat_left), '| total:', len(after_filters))
```

**Insights**
- **15 numeric features are dropped as redundant.** The groups match the EDA:
  - **Transaction totals:** `Total a Pagar`, `Total Liquido` and `avg_current_invoice_value` are ρ ≥ 0.98 with `Total Bruto`, so we keep one of them.
  - **Raw vs. log versions:** `spend_365d`/`log_spend_365d` etc. have ρ = 1.00, as expected (log1p does not change the order of the values). We keep one per pair.
  - **Counts vs. spend over the same window:** for 30 and 90 days, the purchase count is slightly stronger than the spend and ρ > 0.90, so the count is kept. For 180 and 365 days the correlation is below 0.90, so both stay.
  - **`past_90d_purchase_rate` is exactly `purchase_count_90d / 90`** (ρ = 1.00), so it adds nothing new.
- **Categorical:** `commercial_zone` and `country_or_region` are strongly associated (Cramér's V = 0.89). We do not drop either here, because both are judged again in 3.5–3.6.
- **41 candidates remain** (32 numeric, 9 categorical).

### 3.5. Embedded method – L1-regularised logistic regression

The L1 (Lasso) penalty pushes the coefficients of weak features to exactly zero, so the model selects features while it is trained. We fit it on the training part of each of the 4 folds, with 3 penalty strengths (`C`; smaller = stronger penalty). For each original feature we count in how many folds it keeps at least one non-zero coefficient. A feature chosen in **3 or 4 of 4 folds** is a stable choice.

```python
# after one-hot encoding we get many columns, this gives back the original feature of each column
def original_names(ct):
    names = []
    for name, step, cols in ct.transformers_:
        if name == 'remainder':
            continue
        if name == 'numeric':
            imputer = step.named_steps['impute']
            names += list(cols) + [cols[i] for i in imputer.indicator_.features_]
        else:
            encoder = step.named_steps['encode']
            for col, cats in zip(cols, encoder.categories_):
                names += [col] * len(cats)
    return np.array(names)

C_values = [0.02, 0.05, 0.1]
l1_counts = pd.DataFrame(0, index=after_filters, columns=[f'C={c}' for c in C_values])
l1_auc = {f'C={c}': [] for c in C_values}

for F_tr, y_tr, F_va, y_va in fold_data:
    for c in C_values:
        l1 = LogisticRegression(penalty='l1', solver='liblinear', C=c, max_iter=5000, random_state=SEED)
        pipe = make_pipe(after_filters, l1).fit(F_tr, y_tr)
        names = original_names(pipe.named_steps['columns'])
        coefs = pipe.named_steps['model'].coef_[0]
        used = set(names[np.abs(coefs) > 1e-6])
        l1_counts[f'C={c}'] += [int(f in used) for f in after_filters]
        l1_auc[f'C={c}'].append(roc_auc_score(y_va, pipe.predict_proba(F_va)[:, 1]))

for c in l1_counts.columns:
    print(f'{c}: mean AUC {np.mean(l1_auc[c]):.4f}, features kept in 3+ folds: {(l1_counts[c] >= 3).sum()}')
print('random_noise kept in how many folds:', l1_counts.loc['random_noise'].to_dict())

with pd.option_context('display.max_rows', 100):
    display(l1_counts.sort_values(['C=0.05', 'C=0.02', 'C=0.1'], ascending=False))
```

**Insights**
- **Stronger penalties keep fewer features with almost no loss in AUC** (17 stable features at C = 0.02 with AUC 0.780; 28 at C = 0.1 with AUC 0.784). Much of the candidate set is not needed by a linear model.
- **Always kept:** recency/frequency (`purchase_count_365d`, `purchase_count_90d`, `prior_mean_gap_days`, `prior_std_gap_days`), the engineered transaction ratios (`current_profit_margin`, `current_discount_rate`) and the month features.
- **L1 alone is not enough.** At C = 0.05 it keeps `random_noise` in 2 of 4 folds, and at C = 0.1 in 3 of 4. With many correlated inputs, a sparse linear model can pick up noise by chance. That is why we also use the noise benchmark in 3.6.
- **L1 is weak for multi-level categorical variables.** `transport_method`, `payment_type` and `commercial_region` are spread over many one-hot columns, each with a small effect, so the penalty removes them even though the EDA showed clear differences in repeat rates. This is a limitation of the method, not proof that they are useless.

### 3.6. Permutation importance compared with `random_noise`

Permutation importance measures how much the validation ROC-AUC drops when one feature's values are randomly shuffled, which breaks its link with the target. It is measured **on the later block of each fold**, so it rewards features that help predict *future* purchases, not features the model only memorised. We use the gradient-boosting model, which also captures non-linear effects and interactions that L1 misses, and repeat each shuffle 10 times.

**Benchmark:** `random_noise` has no real relation with the target, so its importance shows how large an importance can be **by chance**. Features far above it clearly carry signal. Features close to it are uncertain, so we do not decide on them with a fixed cut-off: the ranking from this step is used in 3.7, where the final cut is chosen by measuring performance.

```python
perm = pd.DataFrame(index=after_filters)
for i, (F_tr, y_tr, F_va, y_va) in enumerate(fold_data):
    pipe = make_pipe(after_filters, clone(hgb_model)).fit(F_tr, y_tr)
    result = permutation_importance(pipe, F_va[after_filters], y_va, scoring='roc_auc',
                                    n_repeats=10, random_state=SEED, n_jobs=-1)
    perm[f'fold_{i+1}'] = result.importances_mean

perm_summary = pd.DataFrame({'mean_auc_drop': perm.mean(axis=1),
                             'min_fold': perm.min(axis=1),
                             'folds_positive': (perm > 0).sum(axis=1)})
perm_summary = perm_summary.sort_values('mean_auc_drop', ascending=False)
noise_level = perm_summary.loc['random_noise', 'mean_auc_drop']
print('random_noise importance:', round(noise_level, 4))

plot_df = perm_summary.iloc[::-1]
colors = ['grey' if f == 'random_noise' else '#2a6fdb' for f in plot_df.index]
fig, ax = plt.subplots(figsize=(9, 10))
ax.barh(plot_df.index, plot_df['mean_auc_drop'], color=colors, height=0.7)
ax.axvline(noise_level, color='dimgrey', ls='--', lw=1.2, label='random_noise level')
ax.axvline(0, color='black', lw=0.8)
ax.set_xlabel('Mean drop in ROC-AUC when the feature is shuffled (4 time folds)')
ax.set_title('Permutation importance (gradient boosting)')
ax.tick_params(axis='y', labelsize=8)
ax.legend(loc='lower right', fontsize=8)
plt.tight_layout()
plt.show()
```

**Insights**
- **The noise level is small (mean AUC drop 0.0013), and 20 features are above it.**
- **Recency and frequency dominate**, as in the EDA (finding #13): `purchase_count_365d` alone causes an AUC drop of 0.028, about 20× the noise level, followed by `purchase_count_90d` and `log_spend_365d`.
- **Some features that L1 dropped are clearly useful here:** `transport_method`, `payment_type`, `commercial_region` and `customer_age_at_purchase`. Their effect is spread over several categories or is non-linear, which a tree model handles better.
- **Several engineered features from Section 2 earn their place:** `current_profit_margin`, `current_discount_rate`, `lateness_ratio`, the log spend features and the month/day-of-week features.
- **Around or below the noise level:** `document_series`, `warehouse`, `commercial_zone`, `country_or_region`, `island`, `salesperson`, `is_first_purchase`, `recency_reconstructed_flag`, `Custo`, `Total Portes`, `Total Desc_ Global` and others.
  - `is_first_purchase` being useless may look surprising (EDA: 33% repeat rate). But first purchases are already identified by other features (e.g. `prior_purchase_count` = 0, and missing history values with their missing indicators).
- **Many features sit very close to the noise level** (between 0.0005 and 0.0025). A fixed rule like "keep if above noise" would be fragile for them: a tiny change in the importance values would add or remove several features. That is why the final cut is made in 3.7.

### 3.7. Backward elimination (wrapper)

We take the 40 features left after the filters (without `random_noise`), ranked from most to least important in 3.6. Then we **remove the weakest feature, re-score the remaining set on the 4 time folds with both models, and repeat**, down to 5 features. This gives a curve of performance against the number of features.

**Rule:** we keep the **smallest set whose mean ROC-AUC (average of LR and HGB) is within 0.002 of the best set**. A difference of 0.002 is far below the variation between folds (std ≈ 0.03) and of the same size as what `random_noise` alone changes (3.8), so sets within this range predict equally well. Choosing the smallest one gives a simpler and more stable model.

```python
# rank the features from most to least important (random_noise is only a reference, so we leave it out)
ranked = [f for f in perm_summary.index if f != 'random_noise']
noise_rank = list(perm_summary.index).index('random_noise')
print('Features above random_noise:', noise_rank)

# backward elimination: remove the weakest feature, check the AUC on the 4 folds, repeat
results = []
for n in range(len(ranked), 4, -1):
    lr_auc, _ = cv_folds(ranked[:n], lr_model)
    hgb_auc, _ = cv_folds(ranked[:n], hgb_model)
    results.append([n, lr_auc.mean(), hgb_auc.mean()])

curve = pd.DataFrame(results, columns=['n_features', 'LR_ROC_AUC', 'HGB_ROC_AUC']).set_index('n_features')
curve['mean_AUC'] = (curve['LR_ROC_AUC'] + curve['HGB_ROC_AUC']) / 2

# keep the smallest set that is within 0.002 of the best mean AUC
best_n = curve['mean_AUC'].idxmax()
best_auc = curve['mean_AUC'].max()
n_chosen = curve.index[curve['mean_AUC'] >= best_auc - 0.002].min()
selected = ranked[:n_chosen]
print(f'Best mean AUC: {best_auc:.4f} with {best_n} features')
print(f'Chosen: {n_chosen} features, mean AUC {curve.loc[n_chosen, "mean_AUC"]:.4f}')

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(curve.index, curve['LR_ROC_AUC'], color='#2a6fdb', lw=1.5, label='Logistic regression')
ax.plot(curve.index, curve['HGB_ROC_AUC'], color='#e8710a', lw=1.5, label='Gradient boosting')
ax.plot(curve.index, curve['mean_AUC'], color='black', lw=2, label='Mean of both')
ax.axvline(n_chosen, color='green', ls='--', lw=1.2, label=f'Chosen: {n_chosen} features')
ax.axvline(noise_rank, color='grey', ls=':', lw=1.2, label='Features above random_noise')
ax.set_xlabel('Number of features kept (most important first)')
ax.set_ylabel('Mean ROC-AUC (4 time folds)')
ax.set_title('Backward elimination')
ax.invert_xaxis()
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

# one line per candidate feature with the reason for keeping or dropping it
decision = pd.DataFrame(index=all_candidates)
decision['L1_folds(C=0.05)'] = l1_counts['C=0.05']
decision['perm_mean'] = perm_summary['mean_auc_drop'].round(4)
reason = {}
for f in all_candidates:
    if f == 'random_noise':
        reason[f] = 'drop: negative control'
    elif f in near_constant.index:
        reason[f] = 'drop: near-constant'
    elif f in list(redundant['dropped']):
        reason[f] = 'drop: redundant with ' + redundant.set_index('dropped').loc[f, 'kept_instead']
    elif f in selected:
        reason[f] = 'KEEP'
    else:
        reason[f] = 'drop: removed by backward elimination'
decision['decision'] = pd.Series(reason)

agree = (decision.loc[selected, 'L1_folds(C=0.05)'] >= 3).sum()
print('Selected features also kept by L1 in 3+ folds:', agree, 'of', len(selected))
with pd.option_context('display.max_rows', 100):
    display(decision.dropna(subset=['perm_mean']).sort_values('perm_mean', ascending=False))
```

**Insights**
- **The curve is flat from 40 down to about 23 features, then falls quickly.** Below 14 features the mean AUC drops by more than 0.005 compared with the chosen set, so the strongest features alone are not enough.
- **The best mean AUC (0.7882) is reached with 35 features, and 23 features are within 0.002 of it (0.7865).** We keep these **23 features**.
- **The cut is close to the noise level, but not identical.** The 20 features above `random_noise` are all kept, plus `days_since_previous_purchase`, `N_ Detalhes` and `customer_tenure_days`, which are just below it: removing them lowers the AUC on the curve, so together they still add information.
- **11 of the 23 selected features are also chosen by L1** in at least 3 folds. The others are mainly the categorical and non-linear features that L1 handles badly (3.5).
- **Removed by backward elimination** (among others): `log_prior_mean_purchase_value`, `past_90d_avg_spend`, `Custo`, `purchase_month_sin`, `salesperson`, `island`, `country_or_region`, `commercial_zone`, `warehouse` and `document_series`.

### 3.8. Ablations

Backward elimination removes features one by one in a fixed order. Here we check the selected set as a whole and answer the questions left open in Sections 1 and 2:
- **Customer block** (finding #21): selected set with vs. without the `clientes.csv` features.
- **Calendar features:** the dates are anonymised (order and gaps preserved), so month and weekday may not be "real" calendar values. We check whether they still help.
- **Distribution shift** (finding #19): `customer_tenure_days` is the variable that changes most between train and test (KS = 0.46). We check what we lose by dropping it.
- **`salesperson` and `document_series`:** `salesperson` was clear in the EDA (finding #15), `document_series` showed no relation; both were removed in 3.7.
- **`random_noise`:** adding it should not help. Its effect shows how big a difference can be **by pure chance**.

Each set is scored with both models on the same 4 folds. Because the folds differ in difficulty, we compare each set **with the selected set fold by fold** (`*_diff`, and in how many of the 4 folds the set did better).

```python
customer_block = ['customer_age_at_purchase', 'commercial_region', 'island', 'country_or_region']
calendar_block = ['purchase_month_sin', 'purchase_month_cos', 'purchase_day_of_week']

feature_sets = {
    'All candidates': all_candidates,
    'After filters (3.3-3.4)': after_filters,
    'Selected set (3.7)': selected,
    'Selected - customer block': [f for f in selected if f not in customer_block],
    'Selected - calendar features': [f for f in selected if f not in calendar_block],
    'Selected - customer_tenure_days': [f for f in selected if f != 'customer_tenure_days'],
    'Selected + salesperson': selected + ['salesperson'],
    'Selected + document_series': selected + ['document_series'],
    'Selected + random_noise': selected + ['random_noise'],
}

fold_auc = {}
rows = []
for name, feats in feature_sets.items():
    lr_auc, lr_ap = cv_folds(feats, lr_model)
    hgb_auc, hgb_ap = cv_folds(feats, hgb_model)
    fold_auc[name] = {'LR': lr_auc, 'HGB': hgb_auc}
    rows.append([name, len(feats), lr_auc.mean(), lr_auc.std(), lr_ap.mean(),
                 hgb_auc.mean(), hgb_auc.std(), hgb_ap.mean()])

comparison = pd.DataFrame(rows, columns=['feature_set', 'n_features', 'LR_ROC_AUC', 'LR_sd', 'LR_AP',
                                         'HGB_ROC_AUC', 'HGB_sd', 'HGB_AP']).set_index('feature_set')

# compare each set with the selected set fold by fold, so the difference between folds cancels out
ref = 'Selected set (3.7)'
for m in ['LR', 'HGB']:
    diff = {n: fold_auc[n][m] - fold_auc[ref][m] for n in feature_sets}
    comparison[f'{m}_diff'] = [diff[n].mean() for n in comparison.index]
    comparison[f'{m}_folds_better'] = [(diff[n] > 0).sum() for n in comparison.index]
display(comparison.round(4))

others = [n for n in feature_sets if n != ref]
fig, ax = plt.subplots(figsize=(9, 4.6))
for shift, m, color, marker, label in [(-0.13, 'LR', '#2a6fdb', 'o', 'Logistic regression'),
                                       (0.13, 'HGB', '#e8710a', 's', 'Gradient boosting')]:
    for i, n in enumerate(others):
        d = fold_auc[n][m] - fold_auc[ref][m]
        ax.plot([d.min(), d.max()], [i + shift, i + shift], color=color, lw=1.2, alpha=0.6)
        ax.plot(d.mean(), i + shift, marker=marker, color=color, ms=6, label=label if i == 0 else None)
ax.axvline(0, color='black', lw=0.9)
ax.set_yticks(range(len(others)))
ax.set_yticklabels([f'{n} (n={len(feature_sets[n])})' for n in others], fontsize=8)
ax.invert_yaxis()
ax.set_xlabel('ROC-AUC difference vs. selected set (dot = mean, line = range over the 4 folds)')
ax.set_title('Compared with the selected set (right of 0 = better)')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.17), ncol=2, fontsize=8, frameon=False)
plt.tight_layout()
plt.show()
```

**Insights**
- **The selected set (23 features) predicts as well as all 79 candidates.** Mean ROC-AUC is 0.784 for LR in both cases, and 0.789 vs. 0.790 for HGB. **More than two thirds of the features can be removed** with a difference of about 0.001, which is the same size as the effect of adding `random_noise` (+0.0009 for HGB).
- **Calendar features clearly help:** without them HGB loses 0.0071 AUC and is worse in **all 4 folds**. Even if the dates were shifted during anonymisation, a constant shift keeps the yearly cycle, so the pattern stays consistent between train and test. We keep `purchase_month_cos` and `purchase_day_of_week`.
- **The customer block helps the tree model:** without it HGB loses 0.0023 AUC (worse in 3 of 4 folds), while LR changes very little (+0.0007). We **keep `customer_age_at_purchase` and `commercial_region`**, but the effect is small.
- **`customer_tenure_days` is kept despite its drift.** Removing it lowers AUC for both models (−0.0011 LR, −0.0027 HGB). It is flagged as a risk for Section 4 (test customers have much longer tenures, median 2,662 vs. 918 days).
- **`salesperson` adds nothing consistent** (+0.0011 LR, −0.0001 HGB). Its EDA effect is probably already captured by `transport_method` and `payment_type`.
- **`document_series` gives a small gain** (+0.0016 LR in 4 of 4 folds, +0.0012 HGB in 3 of 4). This is within the 0.002 tolerance and close to what `random_noise` produces, and the EDA found no relationship with the target (p = 0.97). We leave it out, but Section 4 can retest it on the hold-out.

**Feature-selection decision log**

| Variable(s) | Decision | Reason |
|---|---|---|
| `ID`, `customer_id` | Excluded | Identifiers, not behaviour; `customer_id` would allow memorising customers (3.2) |
| `random_noise` | Dropped | Negative control; used as the "no signal" benchmark (3.6, 3.8) |
| 23 invalid / inconsistency flags | Dropped | Constant or almost constant in training (3.3); invalid values are still handled as missing |
| Raw spend (`spend_*`, `prior_total_spend`, `prior_mean_purchase_value`) | Dropped | ρ ≥ 0.90 with a kept count or log version (3.4) |
| `Total a Pagar`, `Total Liquido`, `avg_current_invoice_value`, `Total Desc_ Linha`, `prior_std_purchase_value` | Dropped | ρ ≥ 0.90 with `Total Bruto` / `current_discount_rate` / `prior_max_purchase_value` (3.4) |
| `past_90d_purchase_rate` | Dropped | Exactly `purchase_count_90d / 90` (3.4) |
| `document_series` | Dropped | No relation in the EDA (p = 0.97), removed by backward elimination; small gain in 3.8 is within the noise range |
| `commercial_zone` | Dropped | Near-zero importance, 97% one level, strong train/test shift and unseen zones in test (finding #17) |
| `warehouse`, `country_or_region`, `island`, `salesperson`, `log_prior_mean_purchase_value`, `past_90d_avg_spend`, `Custo`, `purchase_month_sin` and other weak features | Dropped | Removed by backward elimination: no loss in AUC without them (3.7) |
| `is_first_purchase`, `recency_reconstructed_flag` | Dropped | No added value; first purchases are already identified by the history features |
| Calendar (`purchase_month_cos`, `purchase_day_of_week`) | Kept | Ablation: HGB worse in all 4 folds without them (3.8) |
| Customer block (`customer_age_at_purchase`, `commercial_region`) | Kept | Ablation: HGB worse in 3 of 4 folds without it (3.8), answers finding #21 |
| `customer_tenure_days` | Kept, with caution | Helps both models; strongest train/test drift, to watch in Section 4 |
